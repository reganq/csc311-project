```python
from sentence_transformers import SentenceTransformer
import numpy as np

# 1. Load a pretrained Sentence Transformer model
model = SentenceTransformer("all-MiniLM-L6-v2")

with open('text/clocks_feel.txt', 'r') as f:
    clocks_sentences = [line.rstrip('\n') for line in f]
with open('text/starry_feel.txt', 'r') as f:
    starry_sentences = [line.rstrip('\n') for line in f]
with open('text/lilies_feel.txt', 'r') as f:
    lilies_sentences = [line.rstrip('\n') for line in f]

# 2. Calculate embeddings by calling model.encode()
clocks_embeddings = model.encode(clocks_sentences)
starry_embeddings = model.encode(starry_sentences)
lilies_embeddings = model.encode(lilies_sentences)
vec = model.encode("This one almost makes me want to vomit.")

# 3. Calculate the embedding similarities
clocks_similarity = np.mean(np.array([tensor.item() for tensor in model.similarity(clocks_embeddings, vec)]))
starry_similarity = np.mean(np.array([tensor.item() for tensor in model.similarity(starry_embeddings, vec)]))
lilies_similarity = np.mean(np.array([tensor.item() for tensor in model.similarity(lilies_embeddings, vec)]))

preds = ['clocks', 'starry', 'lilies']
max_similarity = np.max([clocks_similarity, starry_similarity, lilies_similarity])
max_idx = np.argmax([clocks_similarity, starry_similarity, lilies_similarity])
print("Predicted painting:", preds[max_idx], "Similarity:", max_similarity)
```
